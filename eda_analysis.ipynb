```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("C:/Users/sarav/OneDrive/Desktop/retail_sales_project/superstore_sample.csv")
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month
print("Data loaded successfully")
print("Total records:", len(df))

# Which region is making the most sales?
region_sales = df.groupby('Region')['Sales'].sum().sort_values()
region_sales.plot(kind='bar', color='steelblue', figsize=(8,5))
plt.title('Which Region Has Highest Sales?')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.savefig('region_sales.png')
plt.show()

# Which category gives more profit?
category_profit = df.groupby('Category')['Profit'].sum()
category_profit.plot(kind='bar', color='coral', figsize=(8,5))
plt.title('Which Category Is Most Profitable?')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)
plt.savefig('category_profit.png')
plt.show()

# Does giving more discount reduce profit?
plt.figure(figsize=(8,5))
plt.scatter(df['Discount'], df['Profit'], alpha=0.4, color='green')
plt.title('Does Higher Discount Reduce Profit?')
plt.xlabel('Discount Given')
plt.ylabel('Profit Earned')
plt.savefig('discount_vs_profit.png')
plt.show()

# Which months have highest sales?
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.plot(kind='line', marker='o', color='orange', figsize=(10,5))
plt.title('How Did Sales Change Each Month?')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(range(1,13), ['Jan','Feb','Mar','Apr','May','Jun',
                          'Jul','Aug','Sep','Oct','Nov','Dec'])
plt.savefig('monthly_trend.png')
plt.show()

# Who are our best customers?
top_customers = df.groupby('Customer')['Profit'].sum().nlargest(10)
top_customers.plot(kind='barh', color='purple', figsize=(10,5))
plt.title('Top 10 Customers By Profit')
plt.xlabel('Total Profit')
plt.ylabel('Customer Name')
plt.savefig('top_customers.png')
plt.show()

print("All 5 charts created and saved!")
```
